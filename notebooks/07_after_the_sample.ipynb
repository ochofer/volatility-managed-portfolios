# 07. The months after Moreira and Muir's sample: May 2015 to August 2026 (E5)

**Terms used in this notebook.**

| Term | Meaning |
|---|---|
| Factor | the monthly return of one long-short portfolio that captures one source of common movement in stock returns: Mkt-RF, SMB, HML, Mom, RMW and CMA |
| Excess return | a return minus the one-month Treasury bill rate over the same period; every factor is one |
| Realised variance | the variance of a factor computed from the daily returns of one month; Moreira and Muir's definition for the volatility-managed factor (equation 2, p. 1616), Cederburg et al.'s for the real-time strategies (equation 4, p. 98) |
| Volatility-managed factor | the factor's return times c divided by last month's realised variance. Here c is fixed on Moreira and Muir's sample, the months to April 2015, and held after it |
| Real-time combination (S1) | each month, mean-variance weights on the volatility-managed and the original factor, estimated from every earlier month only, turned into one position in the factor and capped at 5 in either direction (notebook 02) |
| Real-time original (S2) | the original factor alone, held each month at its mean over five times its variance, both estimated from every earlier month only, under the same cap |
| Expanding window | every month from the factor's first to the month before the decision |
| Sharpe ratio | the average excess return divided by its standard deviation, annualised: times the square root of 12 |
| Sharpe ratio difference | one strategy's Sharpe ratio minus another's, over the same months |
| Ledoit-Wolf test | Ledoit and Wolf's (2008) test that two return series over the same months have the same Sharpe ratio, with the circular block bootstrap (notebook 06) |
| Interval | the range the test gives for the true Sharpe ratio difference at 95 per cent |
| Half-width | the distance from the estimate to either end of the interval |
| p-value | the probability of a difference at least as far from 0 as the one computed, if the two true Sharpe ratios were equal |
| Median | the middle value of a set ranked by size; of six values, the average of the third and the fourth |
| Draw | one random number, or one bootstrap sample, taken by the random number generator |
| Seed | the number that starts a random number generator, so that the same draws come out on every run |
| Verdict | met or not met: the result of an expectation, read from levels fixed before the code ran |
| Vintage | the version of French's files on the download date, named by the CRSP release they were built from |

**What this notebook does.** It asks the study's own question: what the signal did in the 136 months after Moreira and Muir's sample, which neither paper could have seen. For each factor it compares, with Ledoit and Wolf's test, the real-time combination against the real-time original, the comparison of Cederburg et al.'s Table 5, and, in no verdict, the volatility-managed factor with the paper's c against the original factor. It judges E5.

**Why it comes at this point.** Notebooks 03 to 05 replicated the three papers on their own samples, and notebook 06 checked that the test rejects at about its nominal rate when nothing differs. Both constructions are now carried forward unchanged, into months that played no part in building them.

**The expectation it judges: E5**, fixed in `src/sq/constants.py` on 10 November 2026, before this notebook's code was written. On the months after Moreira and Muir's sample, May 2015 to August 2026 (136 months), with the paper's c fixed on the data to April 2015 and the real-time combination carried forward from its expanding window: for each factor, the Sharpe ratio of the real-time combination against the real-time original S2, as in Cederburg et al.'s Table 5, and of the volatility-managed factor against the original factor itself, each difference with its Ledoit-Wolf interval.

1. On at most two of the six factors the real-time combination's Sharpe ratio exceeds S2's by more than the interval allows: the interval's lower end lies above 0.
2. The median across the six of the Sharpe ratio difference lies within plus or minus the median of the six intervals' half-widths.

E5 is met when both hold. Differences and half-widths are annualised. The same two counts are reported for the volatility-managed factor, in no verdict.

**Why these levels.** The reason fixed with E5 on 10 November, in `src/sq/constants.py`: the study's own question, with the prior the critique gives. Gains an investor could not have set in real time should show little in months neither paper could have known; more than two factors clearing the interval fails the expectation, and the result is reported whichever way it falls.

**What I expect to see, and where it comes from.** E5 met. On their own sample Cederburg et al. found the real-time combination's Sharpe ratio below S2's for four of the six factors, with momentum's difference the only significant one (notebook 04). With 136 months, the intervals are wide: for the volatility-managed factor against the original, notebook 06's intervals on 621 to 1,065 months had half-widths of 0.21 to 0.44, and a sample a fifth to an eighth as long widens them by a factor of about 2 to 3.

```python
%%writefile src/sq/french_loader.py
"""
Ken French data library: download, record, parse, assemble.

The library serves each dataset as a zip holding one CSV. The CSV is not a
single table. It is a text file with one or more blocks, each announced by a title
line ("  Average Value Weighted Returns -- Monthly"), followed by a header row
that starts with a comma, followed by rows that start with a date: eight digits
for daily data (19260701), six for monthly (192607), four for annual (1927).
Missing values are coded -99.99 or -999. Returns are in percent.

This module turns that into pandas DataFrames indexed by a Period (day, month
or year), in decimal returns, with the missing codes replaced by NaN. It also
writes a provenance record for every file it downloads: the URL, the download
time, the SHA-256 of the zip, and the CRSP vintage line French prints at the
top of each CSV. A result is only as reproducible as the vintage it was
computed on, and French revises history, so the record is part of every output.

Two facts about the daily files matter for a monthly variance built from them.
They carry Saturday trading days until 24 May 1952, when the New York Stock
Exchange stopped trading on Saturdays. The number of trading days in a calendar
month therefore runs from 17 to 27 before June 1952 and from 15 to 23 from
1953 on (counted on vintage 202609). The variance definitions in constants.py
say how each paper treats the number of trading days.

Nothing here is committed to the repository: the zips live in a local cache
directory that .gitignore excludes.
"""
from __future__ import annotations

import hashlib
import io
import json
import re
import zipfile
from datetime import datetime, timezone
from pathlib import Path
from zoneinfo import ZoneInfo

import numpy as np
import pandas as pd
import requests

from . import constants as C

_DATE_ROW = re.compile(r"^\s*(\d{8}|\d{6}|\d{4})\s*,")   # daily YYYYMMDD, monthly YYYYMM, annual YYYY


def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()


def download(key: str, cache_dir: str | Path = "data/french", force: bool = False) -> dict:
    """Fetch one French zip by its key in constants.FRENCH_FILES.

    Returns the provenance record for the file and writes the zip to cache_dir.
    A cached zip is reused unless force=True, and its record is rebuilt from the
    cached bytes, so the SHA-256 always describes the file actually on disk.
    """
    filename = C.FRENCH_FILES[key]
    url = C.FRENCH_BASE_URL + filename
    cache_dir = Path(cache_dir)
    cache_dir.mkdir(parents=True, exist_ok=True)
    path = cache_dir / filename
    if force or not path.exists():
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        path.write_bytes(response.content)
        fetched_now = True
    else:
        fetched_now = False
    data = path.read_bytes()
    now_utc = datetime.now(timezone.utc)
    record = {
        "key": key,
        "filename": filename,
        "url": url,
        "fetched_this_run": fetched_now,
        "download_time_utc": now_utc.isoformat(timespec="seconds"),
        "download_time_local": now_utc.astimezone(ZoneInfo(C.TIMEZONE)).isoformat(timespec="seconds"),
        "zip_sha256": sha256_bytes(data),
        "zip_bytes": len(data),
        "vintage_line": _vintage_line(data),
    }
    return record


def _csv_text(zip_bytes: bytes) -> str:
    with zipfile.ZipFile(io.BytesIO(zip_bytes)) as zf:
        names = [n for n in zf.namelist() if n.lower().endswith(".csv")]
        if len(names) != 1:
            raise ValueError(f"expected one CSV in the zip, found {names}")
        return zf.read(names[0]).decode("latin-1")


def _vintage_line(zip_bytes: bytes) -> str:
    """The first line of the CSV, e.g. 'This file was created using the 202607 CRSP database.'"""
    text = _csv_text(zip_bytes)
    for line in text.splitlines():
        if line.strip():
            first = line.strip()
            return first.split(".")[0] + "." if "." in first else first
    return ""


def parse_blocks(text: str) -> list[dict]:
    """Split a French CSV into its blocks.

    Each block is returned as {"title": str, "frequency": "daily" | "monthly" |
    "annual", "frame": DataFrame} where the frame is indexed by Period (day,
    month or year) and holds floats in the file's own units (percent). The title is the last non-empty line before
    the header row that is not itself a data row.
    """
    lines = text.splitlines()
    blocks = []
    i = 0
    n = len(lines)
    while i < n:
        if _DATE_ROW.match(lines[i]):
            # Walk back to the header row (starts with a comma) and the title above it.
            header_idx = i - 1
            while header_idx >= 0 and not lines[header_idx].lstrip().startswith(","):
                header_idx -= 1
            if header_idx < 0:
                raise ValueError(f"data row at line {i} has no header row above it")
            title_idx = header_idx - 1
            while title_idx >= 0 and not lines[title_idx].strip():
                title_idx -= 1
            title = lines[title_idx].strip() if title_idx >= 0 else ""
            # The factor files and the later blocks of the size-and-value file
            # carry no title line, only prose; a sentence is not a title.
            if not title or title.endswith(".") or len(title) > 70:
                title = f"(untitled block {len(blocks)})"
            columns = [c.strip() for c in lines[header_idx].split(",")[1:]]
            rows = []
            j = i
            while j < n and _DATE_ROW.match(lines[j]):
                rows.append(lines[j])
                j += 1
            raw = pd.read_csv(io.StringIO("\n".join(rows)), header=None, index_col=0)
            raw.columns = columns[: raw.shape[1]]
            date_len = len(str(raw.index[0]).strip())
            if date_len == 8:
                idx = pd.PeriodIndex(pd.to_datetime(raw.index.astype(str), format="%Y%m%d"), freq="D")
                freq = "daily"
            elif date_len == 6:
                idx = pd.PeriodIndex([pd.Period(f"{str(d)[:4]}-{str(d)[4:6]}", freq="M") for d in raw.index])
                freq = "monthly"
            else:
                idx = pd.PeriodIndex([pd.Period(str(d), freq="Y") for d in raw.index])
                freq = "annual"
            frame = raw.astype(float)
            frame.index = idx
            frame.index.name = "period"
            blocks.append({"title": title, "frequency": freq, "frame": frame})
            i = j
        else:
            i += 1
    return blocks


def load_blocks(key: str, cache_dir: str | Path = "data/french") -> list[dict]:
    """All blocks of one file, with the file's provenance record attached to each."""
    record = download(key, cache_dir)
    text = _csv_text((Path(cache_dir) / C.FRENCH_FILES[key]).read_bytes())
    blocks = parse_blocks(text)
    for b in blocks:
        b["provenance"] = record
    return blocks


def load_monthly(key: str, block: int | str = 0, cache_dir: str | Path = "data/french",
                 as_decimal: bool = True) -> pd.DataFrame:
    """One block of one file, in decimal returns with missing codes as NaN.

    The name dates from the monthly files; the function serves the daily files
    in the same way (the block's index is then a daily Period).

    block is an integer position among the file's blocks, or a regular
    expression matched against block titles (the first match wins).
    """
    blocks = load_blocks(key, cache_dir)
    if isinstance(block, int):
        chosen = blocks[block]
    else:
        matches = [b for b in blocks if re.search(block, b["title"], flags=re.I)]
        if not matches:
            raise KeyError(f"no block in {key} matches {block!r}; titles: {[b['title'] for b in blocks]}")
        chosen = matches[0]
    frame = cut_to_sample_end(chosen["frame"].copy())
    for code in C.FRENCH_MISSING_CODES:
        frame = frame.mask(np.isclose(frame, code))
    if as_decimal and C.FRENCH_RETURNS_ARE_PERCENT:
        frame = frame / 100.0
    frame.attrs["title"] = chosen["title"]
    frame.attrs["provenance"] = chosen["provenance"]
    return frame


def cut_to_sample_end(frame: pd.DataFrame, end: str | None = None) -> pd.DataFrame:
    """The frame up to the end of the sample (constants.SAMPLE_END): a monthly block to that
    month, a daily block to the last day of that month, an annual block to that year. A later vintage of French's
    files carries months after the end, which the study does not use; the vintage in the provenance record still says
    which revision of history the months before the end came from. With no end set, the frame is returned as it is."""
    end = C.SAMPLE_END if end is None else end
    if end is None or not isinstance(frame.index, pd.PeriodIndex) or len(frame) == 0:
        return frame
    end_m = pd.Period(end, "M")
    freq = frame.index.freqstr or ""
    if freq.startswith("M"):
        limit = end_m
    elif freq.startswith("D"):
        limit = end_m.asfreq("D", "end")
    else:
        limit = end_m.asfreq(frame.index.freq, "end")
    return frame.loc[:limit]


def load_counts(key: str, block: int | str, cache_dir: str | Path = "data/french") -> pd.DataFrame:
    """A block in the file's own units (firm counts, average firm size), missing codes as NaN."""
    return load_monthly(key, block, cache_dir, as_decimal=False)


def write_provenance(records: list[dict], path: str | Path = C.PROVENANCE_FILE) -> Path:
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    ordered = sorted(records, key=lambda r: r["key"])
    path.write_text(json.dumps(ordered, indent=2, sort_keys=True) + "\n")
    return path


def block_summary(key: str, cache_dir: str | Path = "data/french") -> pd.DataFrame:
    """One row per block: title, frequency, columns, rows, first and last period."""
    rows = []
    for k, b in enumerate(load_blocks(key, cache_dir)):
        f = b["frame"]
        rows.append({
            "file": key, "block": k, "title": b["title"], "frequency": b["frequency"],
            "columns": f.shape[1], "rows": f.shape[0],
            "first": str(f.index[0]), "last": str(f.index[-1]),
        })
    return pd.DataFrame(rows)


# ---------------------------------------------------------------------------
# The study's series, assembled from the files above
# ---------------------------------------------------------------------------

def _suffix(freq: str) -> str:
    if freq not in ("monthly", "daily"):
        raise ValueError(f"freq is 'monthly' or 'daily', not {freq!r}")
    return "" if freq == "monthly" else "_daily"


def factors(freq: str = "monthly", cache_dir: str | Path = "data/french") -> pd.DataFrame:
    """The six factors of constants.FACTORS and the risk-free rate, in decimal returns, cut at the sample's end.

    Each factor comes from the file constants.FACTOR_SOURCE names; the columns are aligned on the union of the
    dates, so a factor whose file starts later (momentum, RMW and CMA) is NaN before its first date.
    """
    sfx = _suffix(freq)
    cols = {}
    frames = {}
    for f in C.FACTORS + ("RF",):
        key = (C.FACTOR_SOURCE[f] if f != "RF" else C.RF_SOURCE) + sfx
        if key not in frames:
            frames[key] = load_monthly(key, 0, cache_dir)
        cols[f] = frames[key][f]
    out = pd.DataFrame(cols)
    out.attrs["provenance"] = [frames[k].attrs["provenance"] for k in sorted(frames)]
    return out


def wml(freq: str = "monthly", cache_dir: str | Path = "data/french") -> pd.Series:
    """WML: the top decile minus the bottom decile of the ten value-weighted portfolios formed on momentum,
    in decimal returns, cut at the sample's end (Barroso and Santa-Clara, Appendix A)."""
    key = "mom10" + _suffix(freq)
    frame = load_monthly(key, C.WML_BLOCK, cache_dir)
    if C.WML_BLOCK_TITLE_MUST_CONTAIN not in frame.attrs["title"]:
        raise ValueError(f"block {C.WML_BLOCK} of {key} is {frame.attrs['title']!r}, not the value-weighted returns")
    out = (frame[C.WML_LONG] - frame[C.WML_SHORT]).rename("WML")
    out.attrs["provenance"] = frame.attrs["provenance"]
    return out


def trading_days_per_month(daily: pd.DataFrame | pd.Series) -> pd.Series:
    """The number of daily rows in each calendar month, indexed by month."""
    idx = daily.index
    months = idx.asfreq("M") if isinstance(idx, pd.PeriodIndex) else pd.PeriodIndex(idx, freq="M")
    return pd.Series(1, index=months).groupby(level=0).sum().rename("trading days")
```

```python
%%writefile src/sq/managed.py
"""
The signal and the two constructions built on it.

The signal is last month's realised variance of a factor, computed from the
factor's daily returns. Each paper defines it in its own way, and each
comparison with a paper's printed numbers uses that paper's definition
(constants.RV_BY_COMPARISON):

  rv_mm     Moreira and Muir, equation 2, p. 1616: each day's return minus the
            month's mean daily return, squared and summed over the month.
  rv_cowy   Cederburg, O'Doherty, Wang and Yan, equation 4, p. 98: 22 divided
            by the month's number of trading days J, times the sum of squared
            daily returns.
  rv_bsc    Barroso and Santa-Clara, equation 5, p. 115: 21 times the mean
            squared daily return over the last 126 trading days up to the
            month's last trading day.

The volatility-managed factor (Moreira and Muir, equation 1, p. 1616) holds
c / RV_{t-1} units of the factor in month t, with c set so that the managed
and the original factor have the same standard deviation over the sample.

The real-time combination (Cederburg et al., section 4.3, pp. 106-107) mixes
the volatility-managed and the original factor with mean-variance weights
estimated each month from the months before it only. Its position in the
original factor is y_t = x_sigma c_t / RV_{t-1} + x, truncated at plus or
minus the cap (p. 107: "a leverage constraint of |y_t| <= 5"). The ex post
optimal combination (Table 5, p. 106) holds one pair of weights, set with the
means and the covariance of the same months it is applied to.

Barroso and Santa-Clara's scaled momentum (equation 6, p. 115) holds WML at a
weight of a 12 per cent target volatility over the volatility of the last 126
trading days.

All returns are monthly decimal excess returns indexed by a monthly Period;
daily returns are indexed by a daily Period.
"""
from __future__ import annotations

import numpy as np
import pandas as pd

from . import constants as C


# ---------------------------------------------------------------------------
# The signal
# ---------------------------------------------------------------------------

def _by_month(daily: pd.Series):
    daily = daily.dropna()
    return daily.groupby(daily.index.asfreq("M"))


def rv_mm(daily: pd.Series) -> pd.Series:
    """Moreira and Muir's realised variance of each calendar month (equation 2)."""
    return _by_month(daily).apply(lambda r: float(((r - r.mean()) ** 2).sum())).rename("rv_mm")


def rv_cowy(daily: pd.Series) -> pd.Series:
    """Cederburg et al.'s realised variance of each calendar month (equation 4): 22 / J times the sum of squares."""
    g = _by_month(daily)
    return (C.RV_COWY_DAYS * g.apply(lambda r: float((r ** 2).sum())) / g.size()).rename("rv_cowy")


def rv_bsc(daily: pd.Series) -> pd.Series:
    """Barroso and Santa-Clara's variance at each month's last trading day (equation 5): 21 times the mean
    squared daily return over the last 126 trading days. NaN until 126 trading days are available."""
    daily = daily.dropna()
    mean_sq = (daily ** 2).rolling(C.RV_BSC_TRADING_DAYS, min_periods=C.RV_BSC_TRADING_DAYS).mean()
    month_end = mean_sq.groupby(mean_sq.index.asfreq("M")).last()
    return (C.RV_BSC_DAYS_PER_MONTH * month_end).rename("rv_bsc")


def lagged(rv: pd.Series) -> pd.Series:
    """The variance of month t-1 indexed at month t, the value a position for month t can use."""
    out = rv.copy()
    out.index = out.index + 1
    return out.rename(f"{rv.name}_lag")


# ---------------------------------------------------------------------------
# The volatility-managed factor
# ---------------------------------------------------------------------------

def volatility_managed(f: pd.Series, rv: pd.Series, start: str, end: str, c: float | None = None):
    """The volatility-managed factor on the months start to end.

    Returns (frame, c). The frame has one row per month with the factor's return f, last month's variance
    rv_lag, the position weight = c / rv_lag and the managed return = weight * f. With c None, c is set so
    that the managed and the original return have the same standard deviation over these months (Moreira
    and Muir, p. 1616); a given c is used as it is, for months after the sample c was set on.
    """
    df = pd.DataFrame({"f": f, "rv_lag": lagged(rv)}).loc[start:end].dropna()
    raw = df["f"] / df["rv_lag"]
    if c is None:
        c = float(df["f"].std() / raw.std())
    df["weight"] = c / df["rv_lag"]
    df["managed"] = df["weight"] * df["f"]
    return df, c


# ---------------------------------------------------------------------------
# The real-time combination
# ---------------------------------------------------------------------------

def mean_variance_weights(returns: np.ndarray, gamma: float) -> np.ndarray:
    """(1 / gamma) Sigma^-1 mu for the columns of `returns` (months in rows), with the sample covariance."""
    mu = returns.mean(axis=0)
    sigma = np.atleast_2d(np.cov(returns, rowvar=False))
    return np.linalg.solve(sigma, mu) / gamma


def real_time(f: pd.Series, rv: pd.Series, first: str, end: str, k: int = C.REALTIME_K,
              gamma: float = C.REALTIME_GAMMA, cap: float = C.REALTIME_LEVERAGE_CAP) -> pd.DataFrame:
    """Cederburg et al.'s real-time combination (S1) and real-time original (S2), month by month.

    The months run from `first` to `end`; the first k are the initial training period and carry no
    position. For each later month t the window is every month from `first` to t-1 (an expanding window):

      1. c_t gives the managed and the original factor the same variance over the window;
      2. the means and the covariance of the managed and the original factor over the window give the
         weights (x_sigma, x) = (1 / gamma) Sigma^-1 mu;
      3. the position in the original factor is y_t = x_sigma c_t / RV_{t-1} + x, truncated at +-cap;
      4. S1's return is y_t f_t.

    S2 holds y2_t = (1 / gamma) mu_f / sigma_f^2 of the original factor alone, under the same cap, and
    returns y2_t f_t. Nothing from month t or later enters month t's positions.
    """
    df = pd.DataFrame({"f": f, "rv_lag": lagged(rv)}).loc[first:end].dropna()
    fr = df["f"].to_numpy()
    rv_lag = df["rv_lag"].to_numpy()
    raw = fr / rv_lag
    rows = []
    for i in range(k, len(df)):
        f_win, raw_win = fr[:i], raw[:i]
        c_t = f_win.std(ddof=1) / raw_win.std(ddof=1)
        x_sigma, x = mean_variance_weights(np.column_stack([c_t * raw_win, f_win]), gamma)
        y = x_sigma * c_t / rv_lag[i] + x
        y_capped = float(np.clip(y, -cap, cap))
        y2 = float(np.clip(f_win.mean() / (gamma * f_win.var(ddof=1)), -cap, cap))
        rows.append({
            "month": df.index[i], "window": i, "c": c_t, "x_sigma": x_sigma, "x": x,
            "y_uncapped": y, "y": y_capped, "at_cap": abs(y) > cap,
            "f": fr[i], "s1": y_capped * fr[i], "y2": y2, "s2": y2 * fr[i],
        })
    out = pd.DataFrame(rows).set_index("month")
    out.index = pd.PeriodIndex(out.index, freq="M")
    return out


def ex_post_combination(managed: pd.Series, original: pd.Series, gamma: float = C.REALTIME_GAMMA):
    """Cederburg et al.'s ex post optimal combination, S3 (Table 5, p. 106): one pair of weights
    (x_sigma, x) = (1 / gamma) Sigma^-1 mu from the means and the covariance of the managed and the
    original factor over the months given, held in every one of those months. The weights use the
    months they are applied to, which no investor could have done. No cap is applied: the Table 5 note
    limits the sum of the absolute weights on the two factors to 5, and gamma = 5 keeps these weights
    far below it. The scale of the managed factor (its c) changes x_sigma and leaves the returns as
    they are. Returns (the monthly return series, the weights)."""
    both = pd.concat([managed.rename("managed"), original.rename("original")], axis=1).dropna()
    w = mean_variance_weights(both.to_numpy(), gamma)
    return pd.Series(both.to_numpy() @ w, index=both.index, name="s3"), w


def scaled_momentum(wml: pd.Series, rv: pd.Series, start: str, end: str,
                    target: float = C.BSC_TARGET_VOL_MONTHLY) -> pd.DataFrame:
    """Barroso and Santa-Clara's scaled momentum (equation 6, p. 115) on the months start to end: WML
    times target / sigma_hat, sigma_hat the square root of the variance of equation 5 at the end of the
    month before (rv_bsc, lagged one month), target the monthly volatility of 12 per cent a year. The
    frame holds WML, sigma_hat, the weight target / sigma_hat and the scaled return."""
    sigma = np.sqrt(lagged(rv)).rename("sigma_hat")
    df = pd.concat([wml.rename("wml"), sigma], axis=1).loc[start:end].dropna()
    df["weight"] = target / df["sigma_hat"]
    df["scaled"] = df["weight"] * df["wml"]
    return df


def turnover(position: pd.Series) -> pd.Series:
    """The position's turnover each month: the absolute change from the month before, |w_t - w_(t-1)|.
    The first month carries none. The factor's own trading is the same for the managed and the original
    factor and is left out of both."""
    return position.diff().abs().fillna(0.0).rename("turnover")


def net_of_cost(managed: pd.Series, position: pd.Series, cost: float = C.COST_PER_TURNOVER) -> pd.Series:
    """The managed return less `cost` per unit of the position's turnover in the month (Moreira and Muir,
    Table IV, p. 1625)."""
    return (managed - cost * turnover(position)).rename("net")


def percentiles(s: pd.Series, levels=(5, 50, 95)) -> dict:
    """The given percentiles of a series and its maximum, keyed P5, P50, ..., max."""
    out = {f"P{p}": float(np.percentile(s.dropna(), p)) for p in levels}
    out["max"] = float(s.max())
    return out
```

```python
%%writefile src/sq/regression.py
"""
The regression of a volatility-managed factor on its original factor, and the
measures the papers report from it.

    f_sigma_t = alpha + beta f_t + e_t, monthly, by ordinary least squares.

Two standard errors are computed for every coefficient:

  White (1980), the heteroskedasticity-robust standard error, which allows the
  size of e_t to change from month to month. Moreira and Muir's Table I and
  Cederburg et al.'s Table 3 use it (constants.SE_E1); E1 is judged with it.

  Newey and West (1987), which also allows e_t to be correlated with the
  previous L months' e, with weights 1 - l / (L + 1) on lag l (the Bartlett
  kernel), L = constants.NEWEY_WEST_LAGS. It is reported beside White's and is
  the one E4 to E6 use.

Neither applies a small-sample correction, which is the textbook form of both.

Units follow Moreira and Muir's tables (Table I note, p. 1619): alpha, its
standard error and the RMSE (the standard deviation of e) are multiplied by 12
and given in per cent a year; the appraisal ratio is alpha / RMSE times the
square root of 12 (footnote 11, p. 1620); Sharpe ratios are the monthly mean
over the monthly standard deviation times the square root of 12.

Two Sharpe ratios are compared with Jobson and Korkie's test, as Cederburg et
al. compare them (Table 1, p. 99, and Table 5, p. 106).

The maximum drawdown is the largest fall of compounded wealth from its highest
earlier level, as Barroso and Santa-Clara report it (p. 112).
"""
from __future__ import annotations

from statistics import NormalDist

import numpy as np
import pandas as pd

from . import constants as C


def ols(y: np.ndarray, x: np.ndarray, nw_lags: int = C.NEWEY_WEST_LAGS) -> dict:
    """Regression of y on a constant and x. Returns the coefficients, both standard errors, the residuals,
    N, R2 and the RMSE, all monthly and in decimals."""
    y = np.asarray(y, dtype=float)
    X = np.column_stack([np.ones(len(y)), np.asarray(x, dtype=float)])
    n, k = X.shape
    xtx_inv = np.linalg.inv(X.T @ X)
    coef = xtx_inv @ X.T @ y
    e = y - X @ coef
    xe = X * e[:, None]                                  # row t is x_t e_t
    meat_white = xe.T @ xe
    meat_nw = meat_white.copy()
    for lag in range(1, nw_lags + 1):
        w = 1 - lag / (nw_lags + 1)
        g = xe[lag:].T @ xe[:-lag]
        meat_nw += w * (g + g.T)
    se_white = np.sqrt(np.diag(xtx_inv @ meat_white @ xtx_inv))
    se_nw = np.sqrt(np.diag(xtx_inv @ meat_nw @ xtx_inv))
    ss_res = float(e @ e)
    ss_tot = float(((y - y.mean()) ** 2).sum())
    return {
        "alpha": coef[0], "beta": coef[1],
        "alpha_se_white": se_white[0], "beta_se_white": se_white[1],
        "alpha_se_nw": se_nw[0], "beta_se_nw": se_nw[1],
        "residuals": e, "N": n, "R2": 1 - ss_res / ss_tot,
        "rmse": np.sqrt(ss_res / (n - k)),
    }


def sharpe(r: pd.Series | np.ndarray) -> float:
    """The annualised Sharpe ratio of a monthly excess return: mean over standard deviation times the
    square root of 12."""
    r = np.asarray(r, dtype=float)
    return float(r.mean() / r.std(ddof=1) * C.SHARPE_ANNUALISE)


def max_drawdown(r: pd.Series | np.ndarray) -> float:
    """The largest fall of wealth from its highest earlier level, as a fraction (-0.5 is a fall of half).
    Wealth starts at 1 and grows by (1 + r) each month, the return compounded, with nothing added for
    cash: the cumulative return of one dollar held in the strategy, as Barroso and Santa-Clara state
    theirs (p. 113)."""
    wealth = np.cumprod(1 + np.asarray(r, dtype=float))
    peak = np.maximum.accumulate(np.concatenate([[1.0], wealth]))[1:]
    return float((wealth / peak - 1).min())


def jobson_korkie(r_i: pd.Series | np.ndarray, r_j: pd.Series | np.ndarray) -> dict:
    """Jobson and Korkie's (1981) test that two return series over the same months have the same Sharpe
    ratio, with the correction of Memmel (2003), as Cederburg et al. write it (p. 99):

        z = (s_j m_i - s_i m_j) / sqrt(theta),
        theta = (1 / T) (2 s_i^2 s_j^2 - 2 s_i s_j s_ij + m_i^2 s_j^2 / 2 + m_j^2 s_i^2 / 2
                         - (m_i m_j / (s_i s_j)) s_ij^2),

    m the monthly means, s the monthly standard deviations, s_ij the covariance, T the months. z is
    compared with the normal distribution. Returns the annualised Sharpe ratio difference i minus j, z
    and the two-sided p-value."""
    r_i = np.asarray(r_i, dtype=float)
    r_j = np.asarray(r_j, dtype=float)
    if len(r_i) != len(r_j):
        raise ValueError("the two series must cover the same months")
    t = len(r_i)
    m_i, m_j = r_i.mean(), r_j.mean()
    s_i, s_j = r_i.std(ddof=1), r_j.std(ddof=1)
    s_ij = float(np.cov(r_i, r_j)[0, 1])
    theta = (2 * s_i**2 * s_j**2 - 2 * s_i * s_j * s_ij + 0.5 * m_i**2 * s_j**2 + 0.5 * m_j**2 * s_i**2
             - (m_i * m_j / (s_i * s_j)) * s_ij**2) / t
    z = (s_j * m_i - s_i * m_j) / np.sqrt(theta)
    p = 2 * (1 - NormalDist().cdf(abs(z)))
    return {"difference": (m_i / s_i - m_j / s_j) * C.SHARPE_ANNUALISE, "z": float(z), "p": float(p)}


def table_i_row(managed: pd.Series, original: pd.Series) -> dict:
    """One column of Moreira and Muir's Table I, in the paper's units, with both standard errors."""
    fit = ols(managed.to_numpy(), original.to_numpy())
    a = C.ALPHA_ANNUALISE * 100
    return {
        "beta": fit["beta"], "beta_se_white": fit["beta_se_white"],
        "alpha": fit["alpha"] * a, "alpha_se_white": fit["alpha_se_white"] * a, "alpha_se_nw": fit["alpha_se_nw"] * a,
        "N": fit["N"], "R2": fit["R2"], "RMSE": fit["rmse"] * a,
        "appraisal": fit["alpha"] / fit["rmse"] * C.SHARPE_ANNUALISE,
        "sharpe_managed": sharpe(managed), "sharpe_original": sharpe(original),
    }
```

```python
%%writefile src/sq/verdicts.py
"""
The expectations judged, one function each, from the computed and the
published numbers. Each returns a table with one row per comparison and a
verdict that reads the levels from constants.py and nothing else.
"""
from __future__ import annotations

from statistics import NormalDist

import pandas as pd

from . import constants as C
from . import published as P

# The two-sided 5 per cent line for a t-statistic, from the normal distribution: 1.96.
Z_5 = NormalDist().inv_cdf(1 - C.SIGNIFICANCE_LEVEL / 2)


def e1(computed: pd.DataFrame) -> tuple[pd.DataFrame, bool]:
    """E1 on Moreira and Muir's sample. `computed` has one row per factor with the columns of
    regression.table_i_row. Returns the comparison table, one row per factor, and whether E1 is met."""
    rows = []
    for f in C.FACTORS:
        beta, beta_se, alpha_p, se_p, n_p, r2_p, rmse_p = P.MM_TABLE_I[f]
        c = computed.loc[f]
        t_p = alpha_p / se_p
        t_c = c["alpha"] / c["alpha_se_white"]
        ar_p = P.mm_appraisal_from_table_i(f)
        row = {
            "factor": f,
            "alpha published": alpha_p, "alpha computed": c["alpha"],
            "se published": se_p, "se computed (White)": c["alpha_se_white"], "se computed (Newey-West)": c["alpha_se_nw"],
            "t published": t_p, "t computed (White)": t_c,
            "appraisal published": ar_p, "appraisal computed": c["appraisal"],
            "N published": n_p, "N computed": int(c["N"]),
            "sign matches": (alpha_p > 0) == (c["alpha"] > 0),
            "5% category matches": (abs(t_p) > Z_5) == (abs(t_c) > Z_5),
            "alpha within one published se": abs(c["alpha"] - alpha_p) <= C.E1_ALPHA_WITHIN_SE * se_p,
            "appraisal within 0.05": abs(c["appraisal"] - ar_p) <= C.E1_APPRAISAL_BAND,
        }
        legs = ["sign matches", "5% category matches", "alpha within one published se", "appraisal within 0.05"]
        if f == "Mkt-RF":
            row["Sharpe published (Table V)"] = P.MM_TABLE_V_MARKET["sharpe"]
            row["Sharpe computed"] = c["sharpe_managed"]
            row["Sharpe within 0.05"] = abs(c["sharpe_managed"] - P.MM_TABLE_V_MARKET["sharpe"]) <= C.E1_MARKET_SHARPE_BAND
            legs.append("Sharpe within 0.05")
        row["factor matches"] = all(row[leg] for leg in legs)
        rows.append(row)
    table = pd.DataFrame(rows).set_index("factor")
    met = int(table["factor matches"].sum()) >= C.E1_MIN_FACTORS
    return table, met


def _outside(computed: float, published: float, band: float) -> float:
    """How far a computed number lies outside the band around the published one (0 when inside)."""
    return max(0.0, abs(computed - published) - band)


def e2(computed: pd.DataFrame) -> tuple[pd.DataFrame, dict, bool]:
    """E2 on Cederburg et al.'s sample. `computed` has one row per factor and the columns sharpe_original,
    sharpe_managed (Table 1), s1, s2, s1_minus_s2, p_s1_s2, s3, s1_minus_s3 (Table 5, Panel A), Sharpe
    ratios annualised. Returns the comparison table (one row per factor, four comparisons each), the
    result of each clause (i) to (iv), and whether E2 is met: every clause must hold."""
    band, band_s3 = C.E2_SHARPE_BAND, C.E2_S1_S3_BAND
    rows = []
    for f in C.FACTORS:
        c = computed.loc[f]
        o_p, m_p = P.COWY_TABLE_1_ORIGINAL[f][2], P.COWY_TABLE_1_MANAGED[f][2]
        s1_p, s2_p, d12_p, s3_p, d13_p = P.COWY_TABLE_5[f]
        p12_p = P.COWY_TABLE_5_P[f][0]
        significant = c["p_s1_s2"] < C.SIGNIFICANCE_LEVEL
        row = {
            "factor": f,
            "original SR published": o_p, "original SR computed": c["sharpe_original"],
            "managed SR published": m_p, "managed SR computed": c["sharpe_managed"],
            "(i) both within 0.05": abs(c["sharpe_original"] - o_p) <= band and abs(c["sharpe_managed"] - m_p) <= band,
            "S1 published": s1_p, "S1 computed": c["s1"], "S2 published": s2_p, "S2 computed": c["s2"],
            "(ii) both within 0.05": abs(c["s1"] - s1_p) <= band and abs(c["s2"] - s2_p) <= band,
            "S1-S2 published": d12_p, "S1-S2 computed": c["s1_minus_s2"],
            "p published": p12_p, "p computed": c["p_s1_s2"],
            "sign matches": (d12_p > 0) == (c["s1_minus_s2"] > 0),
            "significant at 5%": significant,
            "significance as published": significant == (f in C.E2_ONLY_SIGNIFICANT),
            "S1-S3 published": d13_p, "S1-S3 computed": c["s1_minus_s3"],
            "(iv) within 0.05": abs(c["s1_minus_s3"] - d13_p) <= band_s3,
        }
        row["(iii) sign and significance as published"] = row["sign matches"] and row["significance as published"]
        row["comparisons holding"] = int(row["(i) both within 0.05"]) + int(row["(ii) both within 0.05"]) \
            + int(row["(iii) sign and significance as published"]) + int(row["(iv) within 0.05"])
        rows.append(row)
    table = pd.DataFrame(rows).set_index("factor")
    significant_set = set(table.index[table["significant at 5%"]])
    clauses = {
        "(i)": bool(table["(i) both within 0.05"].all()),
        "(ii)": bool(table["(ii) both within 0.05"].all()),
        "(iii)": int(table["sign matches"].sum()) >= C.E2_SIGN_MIN_FACTORS and significant_set == set(C.E2_ONLY_SIGNIFICANT),
        "(iv)": bool(table["(iv) within 0.05"].all()),
    }
    return table, clauses, all(clauses.values())


def e2_line(table: pd.DataFrame, clauses: dict) -> str:
    """The verdict of E2 in words, with the arithmetic beside each clause."""
    band = C.E2_SHARPE_BAND
    n_hold, n_all = int(table["comparisons holding"].sum()), 4 * len(table)
    met = all(clauses.values())
    parts = [f"E2: {'MET' if met else 'NOT MET'}. {n_hold} of {n_all} comparisons hold."]

    def misses(pairs):
        out = []
        for f in table.index:
            for label, comp, pub in pairs:
                c, p = table.loc[f, comp], table.loc[f, pub]
                gap = _outside(c, p, band)
                if gap > 0:
                    out.append(f"{f}'s {label} {c:.3f} against {p:.2f} published, {gap:.3f} outside the band of {band}")
        return out

    m1 = misses([("original Sharpe ratio", "original SR computed", "original SR published"),
                 ("managed Sharpe ratio", "managed SR computed", "managed SR published")])
    parts.append("(i) met: every original and managed Sharpe ratio within 0.05 of Table 1." if clauses["(i)"]
                 else "(i) not met: " + "; ".join(m1) + ".")
    m2 = misses([("S1", "S1 computed", "S1 published"), ("S2", "S2 computed", "S2 published")])
    parts.append("(ii) met: every S1 and S2 within 0.05 of Table 5." if clauses["(ii)"]
                 else "(ii) not met: " + "; ".join(m2) + ".")
    n_sign = int(table["sign matches"].sum())
    sig = ", ".join(table.index[table["significant at 5%"]]) or "none"
    parts.append(f"(iii) {'met' if clauses['(iii)'] else 'not met'}: the sign of S1 minus S2 matches for {n_sign} of "
                 f"{len(table)}, at least {C.E2_SIGN_MIN_FACTORS} required; significant at 5 per cent: {sig}; "
                 f"published: {', '.join(C.E2_ONLY_SIGNIFICANT)}.")
    off = [f for f in table.index if not table.loc[f, "(iv) within 0.05"]]
    parts.append("(iv) met: S1 minus S3 within 0.05 of Table 5 for all six." if clauses["(iv)"]
                 else "(iv) not met: S1 minus S3 outside 0.05 for " + ", ".join(off) + ".")
    return "\n".join(parts)


def e3(computed: dict) -> tuple[pd.DataFrame, bool]:
    """E3 on May 1927 to December 2011. `computed` holds the annualised Sharpe ratios sharpe_wml and
    sharpe_scaled and the maximum drawdowns mdd_wml and mdd_scaled as fractions (-0.5 a fall of half).
    Returns the comparison table, one row per series, and whether all four comparisons hold."""
    rows = []
    for name, sr_key, dd_key, sr_p in (("WML", "sharpe_wml", "mdd_wml", P.BSC_TABLE_3_WML[-1]),
                                       ("WML scaled", "sharpe_scaled", "mdd_scaled", P.BSC_TABLE_3_WML_SCALED[-1])):
        dd_p = P.BSC_MAX_DRAWDOWN[name]
        rows.append({
            "series": name,
            "Sharpe published": sr_p, "Sharpe computed": computed[sr_key],
            "Sharpe within 0.05": abs(computed[sr_key] - sr_p) <= C.E3_SHARPE_BAND,
            "maximum drawdown published (%)": dd_p, "maximum drawdown computed (%)": computed[dd_key] * 100,
            "drawdown within five points": abs(computed[dd_key] * 100 - dd_p) <= C.E3_DRAWDOWN_BAND * 100,
        })
    table = pd.DataFrame(rows).set_index("series")
    met = bool(table["Sharpe within 0.05"].all() and table["drawdown within five points"].all())
    return table, met


def e3_line(table: pd.DataFrame, met: bool) -> str:
    """The verdict of E3 in words, with the arithmetic beside each comparison."""
    n = int(table["Sharpe within 0.05"].sum() + table["drawdown within five points"].sum())
    w, s = table.loc["WML"], table.loc["WML scaled"]
    return (f"E3: {'MET' if met else 'NOT MET'}. {n} of 4 comparisons hold. "
            f"Sharpe ratios: WML {w['Sharpe computed']:.3f} against {w['Sharpe published']:.2f}, "
            f"scaled WML {s['Sharpe computed']:.3f} against {s['Sharpe published']:.2f}, band 0.05. "
            f"Maximum drawdowns: WML {w['maximum drawdown computed (%)']:.2f} against {w['maximum drawdown published (%)']:.2f} per cent, "
            f"scaled WML {s['maximum drawdown computed (%)']:.2f} against {s['maximum drawdown published (%)']:.2f} per cent, band five points.")


def e5_counts(results: pd.DataFrame) -> dict:
    """The two counts E5 reads, from one row per factor with the annualised Sharpe ratio difference and its
    bootstrap interval (columns difference, lower, upper, half_width): the factors whose interval lies above
    zero, and whether the median difference lies within plus or minus the median half-width."""
    above = list(results.index[results["lower"] > 0])
    median_difference = float(results["difference"].median())
    median_half_width = float(results["half_width"].median())
    return {"factors above": above, "count above": len(above),
            "median difference": median_difference, "median half-width": median_half_width,
            "median within": abs(median_difference) <= median_half_width}


def e5(real_time: pd.DataFrame) -> tuple[dict, bool]:
    """E5 on the months after Moreira and Muir's sample: the real-time combination against the real-time
    original, one row per factor. Met when at most E5_MAX_FACTORS_ABOVE intervals lie above zero and the
    median difference lies within the median half-width."""
    counts = e5_counts(real_time)
    met = counts["count above"] <= C.E5_MAX_FACTORS_ABOVE and counts["median within"]
    return counts, met


def e5_line(counts: dict, met: bool, managed: dict | None = None) -> str:
    """The verdict of E5 in words, with the counts of the volatility-managed factor beside it, in no verdict."""
    above = ", ".join(counts["factors above"]) or "none"
    text = (f"E5: {'MET' if met else 'NOT MET'}. The real-time combination's Sharpe ratio exceeds S2's by more than "
            f"the interval allows on {counts['count above']} of 6 factors ({above}), at most "
            f"{C.E5_MAX_FACTORS_ABOVE} allowed; the median difference, {counts['median difference']:+.3f}, lies "
            f"{'within' if counts['median within'] else 'outside'} plus or minus the median half-width, "
            f"{counts['median half-width']:.3f}.")
    if managed is not None:
        m_above = ", ".join(managed["factors above"]) or "none"
        text += (f" In no verdict, the volatility-managed factor against the original: above on "
                 f"{managed['count above']} of 6 ({m_above}); median difference {managed['median difference']:+.3f}, "
                 f"{'within' if managed['median within'] else 'outside'} the median half-width, "
                 f"{managed['median half-width']:.3f}.")
    return text


def e6(market_net: dict, market_percentiles: dict, cowy_percentiles: pd.DataFrame) -> tuple[pd.DataFrame, bool]:
    """E6. `market_net` holds the market's volatility-managed alpha net of the base cost and its Newey-West
    standard error, both in per cent a year, on Moreira and Muir's sample. `market_percentiles` holds the
    market's P50, P75, P90 and P99 of the position there. `cowy_percentiles` has one row per factor with
    P50 and P99 of the position on Cederburg et al.'s sample. Returns one row per comparison and whether
    all hold."""
    rows = []
    t = market_net["alpha"] / market_net["se_nw"]
    rows.append({"comparison": "Mkt-RF alpha net of 10 bp, Moreira and Muir's sample", "published": P.MM_TABLE_IV_MARKET["alpha_10bp"],
                 "computed": market_net["alpha"], "level": f"positive and t above {C.E6_MIN_NW_T}",
                 "holds": market_net["alpha"] > 0 and t > C.E6_MIN_NW_T, "detail": f"Newey-West t {t:.2f}"})
    for q in C.E6_MARKET_PERCENTILES:
        band = C.E6_P99_BAND if q == 99 else C.E6_PERCENTILE_BAND
        pub = P.MM_TABLE_V_MARKET[f"P{q}"]
        comp = market_percentiles[f"P{q}"]
        rows.append({"comparison": f"Mkt-RF position P{q}, Moreira and Muir's sample", "published": pub, "computed": comp,
                     "level": f"within {band}", "holds": abs(comp - pub) <= band, "detail": ""})
    for f in C.FACTORS:
        corr, p01, p50, p99 = P.COWY_TABLE_1_PANEL_D[f]
        for label, pub, band in (("P50", p50, C.E6_PERCENTILE_BAND), ("P99", p99, C.E6_P99_BAND)):
            comp = float(cowy_percentiles.loc[f, label])
            rows.append({"comparison": f"{f} position {label}, Cederburg et al.'s sample", "published": pub, "computed": comp,
                         "level": f"within {band}", "holds": abs(comp - pub) <= band, "detail": ""})
    table = pd.DataFrame(rows).set_index("comparison")
    return table, bool(table["holds"].all())


def e6_line(table: pd.DataFrame, met: bool) -> str:
    """The verdict of E6 in words, with every comparison that does not hold named with its arithmetic."""
    n, n_all = int(table["holds"].sum()), len(table)
    text = f"E6: {'MET' if met else 'NOT MET'}. {n} of {n_all} comparisons hold."
    first = table.iloc[0]
    text += f" The market's alpha net of 10 basis points: {first['computed']:.2f} per cent a year, {first['detail']}."
    misses = table[~table["holds"]]
    if len(misses):
        text += " Outside: " + "; ".join(f"{i} {r['computed']:.2f} against {r['published']:.2f} ({r['level']})"
                                          for i, r in misses.iterrows()) + "."
    return text


def _category(significant: bool, value: float) -> str:
    return ("significantly positive" if value > 0 else "significantly negative") if significant else "not significant"


def e4(size_rate: float, factors: pd.DataFrame) -> tuple[pd.DataFrame, dict, bool]:
    """E4. `size_rate` is the size check's rejection rate. `factors` has one row per factor on Moreira and
    Muir's sample with the alpha, its Newey-West t-statistic (t_nw), the Sharpe ratio difference managed
    minus original (difference) and whether its bootstrap interval excludes zero (excludes_zero).

    The second part is read strictly, as the design ruled on 12 November 2026: a factor counts when
    the sign of its Sharpe ratio difference equals the sign of its alpha, and the bootstrap's verdict at 5
    per cent equals the Newey-West t-statistic's (|t| above 1.96 or not). The count by category, each test
    placing the factor as significantly positive, significantly negative or not significant, is returned
    beside it in no verdict. E4 is met when both parts hold."""
    rows = []
    for f, r in factors.iterrows():
        t_sig = abs(r["t_nw"]) > Z_5
        b_sig = bool(r["excludes_zero"])
        rows.append({
            "factor": f, "alpha": r["alpha"], "t (Newey-West)": r["t_nw"], "t significant": t_sig,
            "Sharpe difference": r["difference"], "bootstrap significant": b_sig,
            "signs equal": (r["alpha"] > 0) == (r["difference"] > 0),
            "significance equal": t_sig == b_sig,
            "category (t)": _category(t_sig, r["alpha"]), "category (bootstrap)": _category(b_sig, r["difference"]),
        })
    table = pd.DataFrame(rows).set_index("factor")
    table["counts (strict)"] = table["signs equal"] & table["significance equal"]
    table["same category"] = table["category (t)"] == table["category (bootstrap)"]
    lo, hi = C.E4_SIZE_RANGE
    parts = {
        "size": lo <= size_rate <= hi,
        "agreement": int(table["counts (strict)"].sum()) >= C.E4_MIN_FACTORS,
        "count strict": int(table["counts (strict)"].sum()),
        "count by category": int(table["same category"].sum()),
        "size rate": size_rate,
    }
    return table, parts, parts["size"] and parts["agreement"]


def e4_line(parts: dict, met: bool) -> str:
    """The verdict of E4 in words."""
    lo, hi = C.E4_SIZE_RANGE
    return (f"E4: {'MET' if met else 'NOT MET'}. Size check: {parts['size rate']:.1%} of the simulated null pairs rejected, "
            f"{'inside' if parts['size'] else 'outside'} {lo:.1%} to {hi:.1%}. Agreement of the bootstrap with the "
            f"Newey-West t-statistic in sign and in significance: {parts['count strict']} of 6, at least "
            f"{C.E4_MIN_FACTORS} required. In no verdict, agreement by category: {parts['count by category']} of 6.")
```

```python
%%writefile src/sq/bootstrap.py
"""
Ledoit and Wolf (2008): inference on the difference of two Sharpe ratios
measured on the same months.

    Ledoit, O. and Wolf, M. (2008). Robust performance hypothesis testing with
    the Sharpe ratio. Journal of Empirical Finance 15, 850-859.

The difference Delta = mu_1 / sigma_1 - mu_2 / sigma_2 is a smooth function of
four means, those of r_1, r_2, r_1^2 and r_2^2 (sigma^2 = E[r^2] - mu^2). Its
standard error comes from the delta method: the gradient of the function times
the long-run covariance matrix Psi of the four series, which allows the months
to be correlated with one another and the returns to have fat tails.

  On the sample itself, Psi is the prewhitened Parzen-kernel estimate with
  Andrews' (1991) automatic bandwidth. Ledoit and Wolf propose the prewhitened
  QS kernel of Andrews and Monahan (1992) and report that the prewhitened
  Parzen kernel "yields very similar performance" (section 3.2.2 and its
  footnote 8, p. 854), and that results with it were "virtually identical"; its
  finite support makes it "somewhat more convenient to implement" (Remark 4.1,
  p. 857).

  The circular block bootstrap resamples blocks of consecutive months, wrapping
  from the last month to the first, so that the dependence within a block is
  kept. On each bootstrap sample, Psi is the average outer product of the
  blocks' scaled means, the estimate that matches the resampling.

  The studentised statistic |Delta* - Delta_hat| / se* is computed on every
  bootstrap sample. Its 95th percentile z* gives the symmetric interval
  Delta_hat +/- z* se(Delta_hat), and the share of bootstrap statistics at least
  as large as |Delta_hat| / se(Delta_hat) gives the p-value of a zero
  difference.

The paper chooses the block size by its Algorithm 3.1, a calibration on data
simulated from a model fitted to the sample (p. 854); this study fixes it at 12
months in its constants, and every caller passes it.

The module depends on numpy alone, so that another study can import it
unchanged. Sharpe ratios here are monthly and use the population variance,
E[r^2] - mu^2, as Ledoit and Wolf do; multiply by the square root of 12 to
annualise.
"""
from __future__ import annotations

import numpy as np

__all__ = ["sharpe_difference", "standard_error", "circular_block_indices", "test", "simulate_null_pair", "size_check"]


def _moments(r1: np.ndarray, r2: np.ndarray):
    return r1.mean(), r2.mean(), (r1 ** 2).mean(), (r2 ** 2).mean()


def _delta(mu1, mu2, g1, g2):
    return mu1 / np.sqrt(g1 - mu1 ** 2) - mu2 / np.sqrt(g2 - mu2 ** 2)


def _gradient(mu1, mu2, g1, g2) -> np.ndarray:
    """The gradient of Delta with respect to (mu1, mu2, g1, g2); works on arrays of draws too."""
    s1, s2 = (g1 - mu1 ** 2) ** 1.5, (g2 - mu2 ** 2) ** 1.5
    return np.stack([g1 / s1, -g2 / s2, -0.5 * mu1 / s1, 0.5 * mu2 / s2], axis=-1)


def sharpe_difference(r1, r2) -> float:
    """The monthly Sharpe ratio of r1 minus that of r2, population variances."""
    r1, r2 = np.asarray(r1, dtype=float), np.asarray(r2, dtype=float)
    return float(_delta(*_moments(r1, r2)))


def _parzen(x: float) -> float:
    x = abs(x)
    if x <= 0.5:
        return 1 - 6 * x ** 2 + 6 * x ** 3
    if x <= 1:
        return 2 * (1 - x) ** 3
    return 0.0


def _gamma(v: np.ndarray, j: int) -> np.ndarray:
    """The sample autocovariance matrix at lag j, divided by T."""
    t = len(v)
    return v[j:].T @ v[:t - j] / t


def _ar1(x: np.ndarray):
    """Least-squares AR(1) with an intercept: the coefficient and the residual variance."""
    y, z = x[1:], x[:-1]
    zc, yc = z - z.mean(), y - y.mean()
    rho = float(zc @ yc / (zc @ zc))
    resid = yc - rho * zc
    return rho, float(resid @ resid / len(resid))


def _andrews_alpha(v: np.ndarray) -> float:
    """Andrews' (1991) plug-in quantity for the Parzen kernel, from univariate AR(1) fits, equal weights."""
    num = den = 0.0
    for i in range(v.shape[1]):
        rho, sig2 = _ar1(v[:, i])
        num += 4 * rho ** 2 * sig2 ** 2 / (1 - rho) ** 8
        den += sig2 ** 2 / (1 - rho) ** 4
    return num / den


def _psi_parzen(v: np.ndarray) -> np.ndarray:
    """The Parzen-kernel long-run covariance with Andrews' bandwidth 2.6614 (alpha T)^(1/5)."""
    t, p = v.shape
    bandwidth = 2.6614 * (_andrews_alpha(v) * t) ** 0.2
    psi = _gamma(v, 0)
    j = 1
    while j < bandwidth and j < t:
        g = _gamma(v, j)
        psi = psi + _parzen(j / bandwidth) * (g + g.T)
        j += 1
    return psi * t / (t - p)


def standard_error(r1, r2) -> float:
    """The standard error of the monthly Sharpe ratio difference: the delta method with the prewhitened
    Parzen-kernel estimate of Psi (a VAR(1) fitted to the four series, its singular values held within
    +/- 0.97, the kernel estimate on the residuals, then recoloured)."""
    r1, r2 = np.asarray(r1, dtype=float), np.asarray(r2, dtype=float)
    mu1, mu2, g1, g2 = _moments(r1, r2)
    grad = _gradient(mu1, mu2, g1, g2)
    v = np.column_stack([r1 - mu1, r2 - mu2, r1 ** 2 - g1, r2 ** 2 - g2])
    t = len(v)
    x, y = v[:-1], v[1:]
    a = np.linalg.lstsq(x, y, rcond=None)[0].T                  # y_t = A y_{t-1} + e_t, no intercept
    u, d, vt = np.linalg.svd(a)
    a = u @ np.diag(np.clip(d, -0.97, 0.97)) @ vt
    resid = y - x @ a.T
    inv = np.linalg.inv(np.eye(4) - a)
    psi = inv @ _psi_parzen(resid) @ inv.T
    return float(np.sqrt(grad @ psi @ grad / t))


def circular_block_indices(months: int, block: int, draws: int, rng: np.random.Generator) -> np.ndarray:
    """Indices of `draws` circular block bootstrap samples: floor(months / block) blocks of `block`
    consecutive months each, every block starting at a month drawn uniformly, wrapping past the last
    month to the first. Shape (draws, blocks * block)."""
    n_blocks = months // block
    starts = rng.integers(0, months, size=(draws, n_blocks))
    idx = (starts[:, :, None] + np.arange(block)[None, None, :]) % months
    return idx.reshape(draws, n_blocks * block)


def _bootstrap_statistics(r1, r2, delta_hat, block, draws, rng):
    t = len(r1)
    idx = circular_block_indices(t, block, draws, rng)
    a, b = r1[idx], r2[idx]
    n = idx.shape[1]
    n_blocks = n // block
    mu1, mu2 = a.mean(axis=1), b.mean(axis=1)
    g1, g2 = (a ** 2).mean(axis=1), (b ** 2).mean(axis=1)
    delta = _delta(mu1, mu2, g1, g2)
    y = np.stack([a - mu1[:, None], b - mu2[:, None], a ** 2 - g1[:, None], b ** 2 - g2[:, None]], axis=-1)
    zeta = np.sqrt(block) * y.reshape(draws, n_blocks, block, 4).mean(axis=2)
    psi = np.einsum("dki,dkj->dij", zeta, zeta) / n_blocks
    grad = _gradient(mu1, mu2, g1, g2)
    se = np.sqrt(np.einsum("di,dij,dj->d", grad, psi, grad) / n)
    return np.abs(delta - delta_hat) / se


def test(r1, r2, block: int = 12, draws: int = 2000, level: float = 0.95, seed: int | np.random.Generator = 0) -> dict:
    """Ledoit and Wolf's test of a zero Sharpe ratio difference, r1 minus r2, on paired monthly returns.

    Returns the monthly difference, its standard error, the symmetric studentised interval at `level`, its
    half-width, the critical value z*, the p-value of a zero difference and whether the interval excludes
    zero. The same seed gives the same result."""
    r1, r2 = np.asarray(r1, dtype=float), np.asarray(r2, dtype=float)
    if r1.shape != r2.shape or r1.ndim != 1:
        raise ValueError("two one-dimensional series over the same months are needed")
    rng = seed if isinstance(seed, np.random.Generator) else np.random.default_rng(seed)
    delta_hat = sharpe_difference(r1, r2)
    se = standard_error(r1, r2)
    stats = _bootstrap_statistics(r1, r2, delta_hat, block, draws, rng)
    z_star = float(np.quantile(stats, level))
    d = abs(delta_hat) / se
    p = (int((stats >= d).sum()) + 1) / (draws + 1)
    half = z_star * se
    return {"difference": delta_hat, "se": se, "lower": delta_hat - half, "upper": delta_hat + half,
            "half_width": half, "critical": z_star, "p_value": p, "excludes_zero": bool(abs(delta_hat) > half),
            "months": len(r1), "block": block, "draws": draws, "level": level}


def simulate_null_pair(months: int, mean: float, sd: float, rho: float, dof: int,
                       rng: np.random.Generator) -> tuple[np.ndarray, np.ndarray]:
    """Two monthly return series with the same mean and standard deviation, so the same Sharpe ratio, and
    correlation rho: a common Student-t shock and one own shock each, every shock scaled to variance 1."""
    scale = np.sqrt(dof / (dof - 2))
    common, own1, own2 = (rng.standard_t(dof, size=(3, months)) / scale)
    e1 = np.sqrt(rho) * common + np.sqrt(1 - rho) * own1
    e2 = np.sqrt(rho) * common + np.sqrt(1 - rho) * own2
    return mean + sd * e1, mean + sd * e2


def size_check(pairs: int, months: int, mean: float, sd: float, rho: float, dof: int, seed: int,
               block: int = 12, draws: int = 2000, level: float = 0.95) -> dict:
    """The share of simulated pairs with equal Sharpe ratios whose interval excludes zero: the test's
    rejection rate under the null, which should be close to 1 - level. Returns the rate, the count and
    each pair's difference and p-value."""
    root = np.random.SeedSequence(seed)
    data_seed, boot_seed = root.spawn(2)
    data_rng = np.random.default_rng(data_seed)
    boot_rngs = [np.random.default_rng(s) for s in boot_seed.spawn(pairs)]
    rejected, p_values, differences = [], [], []
    for k in range(pairs):
        r1, r2 = simulate_null_pair(months, mean, sd, rho, dof, data_rng)
        out = test(r1, r2, block=block, draws=draws, level=level, seed=boot_rngs[k])
        rejected.append(out["excludes_zero"])
        p_values.append(out["p_value"])
        differences.append(out["difference"])
    rejected = np.array(rejected)
    return {"rate": float(rejected.mean()), "rejected": int(rejected.sum()), "pairs": pairs,
            "p_values": np.array(p_values), "differences": np.array(differences)}
```

## 4. Ledoit and Wolf's test on both comparisons

Each test draws 2,000 bootstrap samples of 11 blocks of 12 months, 132 of the 136 months, from its own stream of the constants' seed. Differences, intervals and half-widths annualised.

```python
seeds_rt = dict(zip(C.FACTORS, np.random.SeedSequence(C.BOOTSTRAP_SEED).spawn(len(C.FACTORS))))
seeds_vm = dict(zip(C.FACTORS, np.random.SeedSequence(C.BOOTSTRAP_SEED + 1).spawn(len(C.FACTORS))))


def lw_row(x, y, seed):
    t = B.test(x, y, block=C.BOOTSTRAP_BLOCK_MONTHS, draws=C.BOOTSTRAP_DRAWS, level=C.BOOTSTRAP_LEVEL, seed=np.random.default_rng(seed))
    return {"difference": t["difference"] * ANNUAL, "lower": t["lower"] * ANNUAL, "upper": t["upper"] * ANNUAL,
            "half_width": t["half_width"] * ANNUAL, "p_value": t["p_value"]}


rt_tests = pd.DataFrame({f: lw_row(a["s1"], a["s2"], seeds_rt[f]) for f, a in after.items()}).T
vm_tests = pd.DataFrame({f: lw_row(a["managed"], a["original"], seeds_vm[f]) for f, a in after.items()}).T
print("Real-time combination S1 against real-time original S2:")
print(rt_tests.round(3).to_string())
print()
print("Volatility-managed factor (c to April 2015) against the original factor, in no verdict:")
print(vm_tests.round(3).to_string())
```

What could be wrong, and where each point is tested:

- **136 months is a short sample.** The intervals in section 4 are wide, and a difference that lies inside one is not evidence that the two Sharpe ratios are equal; it is a difference these months cannot tell from zero.
- **One period.** The months after the sample hold the calm year of 2017, the crash of March 2020 and the falls of 2022. A different stretch of the same length could give differences of either sign; the sign of each says nothing about another sample.
- **c held from April 2015.** The volatility-managed factor uses a c that an investor in May 2015 could have known; its scale differs from one set on the 136 months, which changes no Sharpe ratio.
- **The vintage.** French's revision of RMW (notebook 04) touches these months as well; the record written at each download names the files used and their checksums.
